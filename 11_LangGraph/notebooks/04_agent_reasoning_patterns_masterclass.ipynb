### Chain-of-Thought (CoT)

CoT asks the model to write intermediate reasoning steps before the final answer, which dramatically improves math and logic tasks compared to answering directly. It is a **prompt technique**, not an agent loop — one forward pass, no tools, no memory.

- **Paper:** [Wei et al., 2022](https://arxiv.org/abs/2201.11903)

![Chain-of-Thought (CoT)](./assets/patterns/cot_architecture.png)

---

### ReAct

ReAct interleaves **Thought → Action → Observation** so the model reasons about what to do, calls a tool, reads the result, and repeats until it can answer. It is the **baseline agent pattern** — flexible but expensive because every tool call needs another LLM invocation.

- **Paper:** [Yao et al., 2022 (ICLR 2023)](https://arxiv.org/abs/2210.03629) | [Tutorial](https://langchain-ai.github.io/langgraph/tutorials/introduction/)

![ReAct](./assets/patterns/react_architecture.png)

![ReAct LangGraph nodes](./assets/patterns/react_langgraph.png)

---

### Reflection / Self-Refine

Reflection runs a **generate → critique → revise** loop where a second LLM pass scores the draft and sends specific feedback back to the generator. No external tools are required — the critic is purely verbal, making this ideal for polishing text, UX copy, or code style.

- **Paper:** [Madaan et al., 2023](https://arxiv.org/abs/2303.17651) | [Tutorial](https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/reflection/reflection.ipynb)

![Reflection / Self-Refine](./assets/patterns/reflection_architecture.png)

![Reflection / Self-Refine LangGraph nodes](./assets/patterns/reflection_langgraph.png)

---

### Reflexion

Reflexion adds **environment feedback** (tool results, test failures) plus **verbal self-reflection** stored in episodic memory, then retries the whole task with that memory. It boosted HumanEval coding accuracy from 80% to 91% without fine-tuning — learning from failure in a single session.

- **Paper:** [Shinn et al., 2023](https://arxiv.org/abs/2303.11366) | [Tutorial](https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/reflexion/reflexion.ipynb)

![Reflexion](./assets/patterns/reflexion_architecture.png)

![Reflexion LangGraph nodes](./assets/patterns/reflexion_langgraph.png)

---

### CRITIC

CRITIC verifies each LLM output using **external tools** (search, calculator, code interpreter) in a verify-then-correct loop — not just asking the model 'are you sure?' Research shows intrinsic self-correction often fails; CRITIC's gains come from **ground truth the model cannot fake**.

- **Paper:** [Gou et al., ICLR 2024](https://arxiv.org/abs/2305.11738)

![CRITIC](./assets/patterns/critic_architecture.png)

![CRITIC LangGraph nodes](./assets/patterns/critic_langgraph.png)

---

### REWOO

REWOO **decouples planning from observation**: the planner writes all tool steps upfront with placeholders (#E1, #E2), workers execute them, then one solver LLM call produces the final answer. It uses ~5× fewer tokens than ReAct because the LLM never re-reads every tool output between steps.

- **Paper:** [Binfeng Xu et al., 2023](https://arxiv.org/abs/2305.18323) | [Tutorial](https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/rewoo/rewoo.ipynb)

![REWOO](./assets/patterns/rewoo_architecture.png)

![REWOO LangGraph nodes](./assets/patterns/rewoo_langgraph.png)

---

### Plan-and-Solve

Plan-and-Solve is a zero-shot prompt that tells the model to **first devise a plan**, then execute it step by step — reducing missing-step errors in multi-step reasoning. It is a prompt pattern; LangGraph's Plan-and-Execute adds explicit graph nodes and replanning.

- **Paper:** [Wang, Lei et al., ACL 2023](https://arxiv.org/abs/2305.04091)

![Plan-and-Solve](./assets/patterns/plan_execute_architecture.png)

---

### Plan-and-Execute

A **planner** LLM writes a multi-step plan; **executor** node(s) run each step with tools; a **replanner** fires if the world changed or a step failed. Production teams use this for SOPs, incident runbooks, and onboarding workflows where step order matters.

- **Planning inspiration: Wang et al., Plan-and-Solve (ACL 2023):** [Wang, Lei et al.](https://arxiv.org/abs/2305.04091)
- **BabyAGI:** [https://github.com/yoheinakajima/babyagi](https://github.com/yoheinakajima/babyagi)
- **LangGraph blog:** [https://blog.langchain.com/planning-agents/](https://blog.langchain.com/planning-agents/)
- **LangGraph tutorial:** [https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/plan-and-execute/plan-and-execute.ipynb](https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/plan-and-execute/plan-and-execute.ipynb)

![Plan-and-Execute](./assets/patterns/plan_execute_architecture.png)

![Plan-and-Execute LangGraph nodes](./assets/patterns/plan_execute_langgraph.png)

### Tree of Thoughts (ToT)

ToT treats reasoning as **tree search**: expand multiple candidate 'thoughts', score each with an LLM judge, prune to top-K, and repeat until a solution is found. Best for problems with a **search space** (puzzles, architecture decisions) — expensive but explores alternatives ReAct misses.

- **Paper:** [Yao et al., NeurIPS 2023](https://arxiv.org/abs/2305.10601) | [Tutorial](https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/tot/tot.ipynb)

![Tree of Thoughts (ToT)](./assets/patterns/tot_architecture.png)

![Tree of Thoughts (ToT) LangGraph nodes](./assets/patterns/tot_langgraph.png)

![Tree of Thoughts (ToT) official LangGraph diagram](./assets/patterns/tot_langgraph.png)

---

### Graph of Thoughts (GoT)

GoT generalizes ToT from a tree to an arbitrary **DAG** — thoughts can merge, backtrack, and combine from different branches. Use when sub-problems solved independently need to be **fused** into one answer (e.g., multi-document synthesis).

- **Paper:** [Besta et al., 2023](https://arxiv.org/abs/2308.09687)

![Graph of Thoughts (GoT)](./assets/patterns/got_architecture.png)

---

### LATS (Language Agent Tree Search)

LATS combines **Monte Carlo Tree Search** with LLM reflection — select a node, expand parallel actions, score with reflection, backpropagate rewards up the tree. It unifies ReAct, Reflexion, ToT, and plan-and-execute into one search algorithm — powerful but the highest cost pattern.

- **Paper:** [Zhou et al., 2023](https://arxiv.org/abs/2310.04406) | [Tutorial](https://blog.langchain.com/reflection-agents/)

![LATS (Language Agent Tree Search)](./assets/patterns/lats_architecture.png)

---

### Self-Discover

Self-Discover **selects reasoning modules** from a library (critical thinking, step-by-step planning, etc.), adapts them to the task, builds a structured JSON plan, then executes it. It is **meta-reasoning** — the model composes its own thinking strategy before solving, ideal for diverse or unfamiliar task types.

- **Paper:** [Zhou et al., 2024](https://arxiv.org/abs/2402.03620) | [Tutorial](https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/self-discover/self-discover.ipynb)

![Self-Discover](./assets/patterns/self_discover_architecture.png)

![Self-Discover LangGraph nodes](./assets/patterns/self_discover_langgraph.png)

---

### Self-RAG

Self-RAG trains/reflection-tokens decide **when to retrieve**, whether the generation is **grounded**, and whether the answer is **useful** — retrieve again or ship. Your Module 10 capstone (`grade_documents` → `rewrite_query` → `check_groundedness`) is a domain-specific Self-RAG graph.

- **Paper:** [Asai et al., 2023](https://arxiv.org/abs/2310.11511) | [Tutorial](https://blog.langchain.com/agentic-rag-with-langgraph/)

![Self-RAG](./assets/patterns/self_rag_architecture.png)

---

### LLM Compiler

LLM Compiler plans a **DAG of tool calls** and executes independent nodes **in parallel**, joining results as dependencies resolve — like a task scheduler for LLM tools. Fastest pattern when tools don't depend on each other's output sequentially (e.g., fetch weather + stock price + news simultaneously).

- **Paper:** [Kim et al., 2023](https://arxiv.org/abs/2312.04511) | [Tutorial](https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/llm-compiler/LLMCompiler.ipynb)

![LLM Compiler](./assets/patterns/llm_compiler_architecture.png)

## Example use case in this notebook

**Support ticket — check CRM account, send password reset if needed**

## Our LangGraph implementation (you will run this next)


```python
from typing import Annotated, TypedDict
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
from langchain_core.tools import tool

# 👇 Mock CRM — deterministic so students see predictable traces
CRM = {
    "user@acme.com": {"exists": True, "plan": "Pro", "reset_sent": False},
    "newhire@acme.com": {"exists": False, "plan": None, "reset_sent": False},
}

@tool
def lookup_account(email: str) -> str:
    """Check if a user account exists in CRM."""
    rec = CRM.get(email.strip().lower())
    if not rec:
        return f"No account found for {email}"
    return f"Account exists. Plan={rec['plan']}, password_reset_sent={rec['reset_sent']}"

@tool
def send_password_reset(email: str) -> str:
    """Send a password reset email to the user."""
    rec = CRM.get(email.strip().lower())
    if not rec or not rec["exists"]:
        return "Cannot send reset — account not found"
    rec["reset_sent"] = True
    return f"Password reset email sent to {email}"

tools = [lookup_account, send_password_reset]
llm_with_tools = llm.bind_tools(tools)

class ReActState(TypedDict):
    messages: Annotated[list, add_messages]

def agent_node(state: ReActState):
    return {"messages": [llm_with_tools.invoke(state["messages"])]}

def should_continue(state: ReActState):
    last = state["messages"][-1]
    if getattr(last, "tool_calls", None):
        return "tools"
    return END

from langgraph.prebuilt import ToolNode
tool_node = ToolNode(tools)

react_builder = StateGraph(ReActState)
react_builder.add_node("agent", agent_node)
react_builder.add_node("tools", tool_node)
react_builder.add_edge(START, "agent")
react_builder.add_conditional_edges("agent", should_continue, ["tools", END])
react_builder.add_edge("tools", "agent")
react_graph = react_builder.compile()
show_graph(react_graph, "ReAct")

TASK = (
    "User user@acme.com cannot log in. Check their account and send a password reset if needed. "
    "Summarize what you did in one sentence."
)
print("Task:", TASK)
result = react_graph.invoke({"messages": [HumanMessage(content=TASK)]})
print("\nFinal:", result["messages"][-1].content)
```

## Example use case in this notebook

**Rewrite a scary NullPointerException stack trace for customers**

## Our LangGraph implementation (you will run this next)


```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

MAX_REFLECTIONS = 2

class ReflectionState(TypedDict):
    raw_error: str
    draft: str
    feedback: str
    reflections: int
    done: bool

# 👇 Realistic bad error from a production API
BAD_ERROR = (
    "HTTP 500 Internal Server Error\n"
    "NullPointerException at com.api.UserService.getProfile(UserService.java:42)\n"
    "Correlation-ID: req_8f3a2b"
)

def generate_node(state: ReflectionState) -> dict:
    if state.get("feedback"):
        prompt = (
            f"Rewrite this error for a non-technical customer.\n"
            f"Original technical error:\n{state['raw_error']}\n\n"
            f"Editor feedback to fix:\n{state['feedback']}\n\n"
            "Write 2-3 sentences. No stack traces. Include correlation ID in plain language."
        )
    else:
        prompt = f"Rewrite this error for a non-technical customer:\n{state['raw_error']}"
    draft = llm.invoke([HumanMessage(content=prompt)]).content
    print(f"  [generate] {draft[:100]}...")
    return {"draft": draft}

def reflect_node(state: ReflectionState) -> dict:
    prompt = (
        "You are a UX writer. Score the customer message 1-10.\n"
        "Criteria: no jargon, actionable next step, empathetic tone.\n"
        "If score >= 8, reply EXACTLY: APPROVED\n"
        "Else bullet critique only.\n\n"
        f"DRAFT:\n{state['draft']}"
    )
    feedback = llm.invoke([HumanMessage(content=prompt)]).content
    done = feedback.strip().upper().startswith("APPROVED")
    print(f"  [reflect] approved={done}")
    return {"feedback": feedback, "reflections": state["reflections"] + 1, "done": done}

def route_reflection(state: ReflectionState) -> str:
    if state["done"] or state["reflections"] >= MAX_REFLECTIONS:
        return "finish"
    return "generate"

def finish_node(state: ReflectionState) -> dict:
    return {}

reflection_builder = StateGraph(ReflectionState)
reflection_builder.add_node("generate", generate_node)
reflection_builder.add_node("reflect", reflect_node)
reflection_builder.add_node("finish", finish_node)
reflection_builder.add_edge(START, "generate")
reflection_builder.add_edge("generate", "reflect")
reflection_builder.add_conditional_edges("reflect", route_reflection, {"generate": "generate", "finish": "finish"})
reflection_builder.add_edge("finish", END)
reflection_graph = reflection_builder.compile()
show_graph(reflection_graph, "Reflection")

out = reflection_graph.invoke({"raw_error": BAD_ERROR, "draft": "", "feedback": "", "reflections": 0, "done": False})
print("\n=== CUSTOMER-FACING MESSAGE ===\n", out["draft"])
```

**What to look at:** No tools in the loop — pure verbal critique. That's the line between Reflection and Reflexion.

**HR policy Q&A — must cite whether dental covers dependents**

## Our LangGraph implementation (you will run this next)


```python
from pydantic import BaseModel, Field

POLICY_KB = {
    "dental": "Pro plan: dental coverage included for employee only. Dependents: vision only, NOT dental.",
    "dependent": "Dependents covered under Pro: medical + vision. Dental is employee-only.",
    "pro plan": "Pro plan includes: medical, dental (employee), vision (employee + dependents).",
}

@tool
def search_policy(query: str) -> str:
    """Search company HR policy documents."""
    q = query.lower()
    hits = [v for k, v in POLICY_KB.items() if k in q]
    return " | ".join(hits) if hits else "NOT FOUND — try dental, dependent, or pro plan"

class RevisionPlan(BaseModel):
    critique: str
    missing_fact: str
    improved_answer: str
    citations: list[str] = Field(default_factory=list)

class ReflexionState(TypedDict):
    question: str
    draft: str
    observation: str
    answer: str
    memory: list
    attempts: int
    done: bool

MAX_REFLEXION = 2
structured_llm = llm.with_structured_output(RevisionPlan)

def draft_node(state: ReflexionState) -> dict:
    mem = "\n".join(state.get("memory") or [])
    prompt = f"Answer briefly (may be wrong — you will verify next):\nQ: {state['question']}\nPast mistakes:\n{mem}"
    draft = llm.invoke([HumanMessage(content=prompt)]).content
    print("  [draft]", draft[:90], "...")
    return {"draft": draft, "attempts": state["attempts"] + 1}

def search_node(state: ReflexionState) -> dict:
    obs = search_policy.invoke({"query": state["question"]})
    print("  [search]", obs)
    return {"observation": obs}

def revise_node(state: ReflexionState) -> dict:
    plan = structured_llm.invoke([
        SystemMessage(content="Improve answer using policy search. Cite policy text."),
        HumanMessage(content=(
            f"Q: {state['question']}\nDraft: {state['draft']}\nPolicy: {state['observation']}"
        )),
    ])
    memory = list(state.get("memory") or [])
    memory.append(f"Attempt {state['attempts']}: {plan.critique}")
    done = "NOT FOUND" not in state["observation"] and "NOT" not in plan.improved_answer.upper()[:20]
    done = done or state["attempts"] >= MAX_REFLEXION
    print("  [revise] done=", done)
    return {"answer": plan.improved_answer, "memory": memory, "done": done}

def route_reflexion(state: ReflexionState) -> str:
    if state["done"] or state["attempts"] >= MAX_REFLEXION:
        return "finish"
    return "draft"

reflexion_builder = StateGraph(ReflexionState)
reflexion_builder.add_node("draft", draft_node)
reflexion_builder.add_node("search", search_node)
reflexion_builder.add_node("revise", revise_node)
reflexion_builder.add_node("finish", lambda s: {})
reflexion_builder.add_edge(START, "draft")
reflexion_builder.add_edge("draft", "search")
reflexion_builder.add_edge("search", "revise")
reflexion_builder.add_conditional_edges("revise", route_reflexion, {"draft": "draft", "finish": "finish"})
reflexion_builder.add_edge("finish", END)
reflexion_graph = reflexion_builder.compile()
show_graph(reflexion_graph, "Reflexion")

HR_Q = "Does our Pro plan cover dental for dependents?"
ref_out = reflexion_graph.invoke({
    "question": HR_Q, "draft": "", "observation": "", "answer": "",
    "memory": [], "attempts": 0, "done": False,
})
print("\n=== FINAL (grounded) ===\n", ref_out["answer"])
```

**What to look at:** `memory` list stores verbal reflections across attempts — the Reflexion paper's episodic buffer.

# Section 6 — CRITIC

**Runnable below ↓**

## What it is (2 sentences)

CRITIC verifies each LLM output using **external tools** (search, calculator, code interpreter) in a verify-then-correct loop — not just asking the model 'are you sure?' Research shows intrinsic self-correction often fails; CRITIC's gains come from **ground truth the model cannot fake**.

## Paper & links

- **Paper:** [Gou et al., ICLR 2024](https://arxiv.org/abs/2305.11738)
- **arXiv:** https://arxiv.org/abs/2305.11738

## Paper architecture diagram

![CRITIC architecture](./assets/patterns/critic_architecture.png)


**LangGraph node graph (matches runnable code below):**

![CRITIC LangGraph nodes](./assets/patterns/critic_langgraph.png)

## Example use case in this notebook

**Finance KPI — verify MRR growth with calculator, not LLM math**

## Our LangGraph implementation (you will run this next)


```python
@tool
def calculator(expression: str) -> str:
    """Evaluate math. Example: 2100000 * 1.12"""
    allowed = set("0123456789+-*/(). ")
    if not set(expression) <= allowed:
        return "Error: invalid chars"
    try:
        return str(round(eval(expression, {"__builtins__": {}}, {}), 2))
    except Exception as e:
        return f"Error: {e}"

class CriticState(TypedDict):
    question: str
    answer: str
    expression: str
    verified_value: str
    rounds: int
    done: bool

MAX_CRITIC = 2

def answer_node(state: CriticState) -> dict:
    prompt = (
        f"Answer: {state['question']}\n"
        "Also output the Python math expression to verify (e.g. 2100000 * 1.12). "
        "Format: ANSWER: ...\nEXPR: ..."
    )
    text = llm.invoke([HumanMessage(content=prompt)]).content
    expr = ""
    for line in text.splitlines():
        if line.upper().startswith("EXPR:"):
            expr = line.split(":", 1)[1].strip()
    print("  [answer]", text[:120], "...")
    return {"answer": text, "expression": expr, "rounds": state["rounds"] + 1}

def verify_node(state: CriticState) -> dict:
    val = calculator.invoke({"expression": state["expression"]}) if state["expression"] else "no expression"
    print("  [verify]", state["expression"], "=", val)
    return {"verified_value": val}

def correct_node(state: CriticState) -> dict:
    prompt = (
        f"Question: {state['question']}\n"
        f"Your answer: {state['answer']}\n"
        f"Calculator says: {state['verified_value']}\n"
        "Fix the answer to match the calculator. Include the correct dollar amount."
    )
    fixed = llm.invoke([HumanMessage(content=prompt)]).content
    return {"answer": fixed, "done": True}

def route_critic(state: CriticState) -> str:
    if state.get("done") or state["rounds"] >= MAX_CRITIC:
        return "finish"
    if state.get("verified_value") and not state["verified_value"].startswith("Error"):
        return "finish"
    return "correct"

critic_builder = StateGraph(CriticState)
critic_builder.add_node("answer", answer_node)
critic_builder.add_node("verify", verify_node)
critic_builder.add_node("correct", correct_node)
critic_builder.add_node("finish", lambda s: {})
critic_builder.add_edge(START, "answer")
critic_builder.add_edge("answer", "verify")
critic_builder.add_conditional_edges("verify", route_critic, {"correct": "correct", "finish": "finish"})
critic_builder.add_edge("correct", "finish")
critic_builder.add_edge("finish", END)
critic_graph = critic_builder.compile()
show_graph(critic_graph, "CRITIC")

FIN_Q = "Our MRR was $2.1 million last quarter and grew 12%. What is the new MRR?"
critic_out = critic_graph.invoke({
    "question": FIN_Q, "answer": "", "expression": "", "verified_value": "", "rounds": 0, "done": False,
})
print("\n=== VERIFIED ANSWER ===\n", critic_out["answer"])
print("Calculator:", critic_out.get("verified_value"))
```

**What to look at:** Calculator provides **ground truth** the model cannot argue with — core CRITIC insight.

# Section 7 — REWOO

**Runnable below ↓**

## What it is (2 sentences)

REWOO **decouples planning from observation**: the planner writes all tool steps upfront with placeholders (#E1, #E2), workers execute them, then one solver LLM call produces the final answer. It uses ~5× fewer tokens than ReAct because the LLM never re-reads every tool output between steps.

## Paper & links

- **Paper:** [Binfeng Xu et al., 2023](https://arxiv.org/abs/2305.18323)
- **arXiv:** https://arxiv.org/abs/2305.18323
- **LangGraph tutorial:** [https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/rewoo/rewoo.ipynb](https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/rewoo/rewoo.ipynb)

## Paper architecture diagram

![REWOO architecture](./assets/patterns/rewoo_architecture.png)


**LangGraph node graph (matches runnable code below):**

![REWOO LangGraph nodes](./assets/patterns/rewoo_langgraph.png)

## Example use case in this notebook

**Analyst question — signups × conversion rate = paid customers**

## Our LangGraph implementation (you will run this next)


```python
import re

@tool
def calc(expression: str) -> str:
    """Safe math eval."""
    allowed = set("0123456789+-*/(). ")
    if not set(expression) <= allowed:
        return "Error"
    return str(eval(expression, {"__builtins__": {}}, {}))

METRICS = {"signups_last_month": "847", "conversion_rate": "0.18", "conversion_delta_pp": "3.2"}

@tool
def lookup_metric(name: str) -> str:
    """Look up a business metric by name."""
    for k, v in METRICS.items():
        if name.lower().replace(" ", "_") in k or k in name.lower():
            return v
    return "NOT FOUND"

class RewooState(TypedDict):
    question: str
    plan: str
    evidence: dict
    answer: str

def plan_node(state: RewooState) -> dict:
    prompt = (
        "Write a REWOO plan using #E1, #E2 placeholders.\n"
        "Tools: lookup_metric[name], calc[expression]\n"
        "Example:\n#E1 = lookup_metric[signups_last_month]\n#E2 = calc[847 * 0.18]\n"
        f"Question: {state['question']}"
    )
    plan = llm.invoke([HumanMessage(content=prompt)]).content
    print("  [plan]\n", plan)
    return {"plan": plan, "evidence": {}}

def worker_node(state: RewooState) -> dict:
    evidence = dict(state.get("evidence") or {})
    for line in state["plan"].splitlines():
        m = re.match(r"\s*(#E\d+)\s*=\s*(lookup_metric|calc|search_policy)\[(.*)\]\s*$", line.strip(), re.I)
        if not m:
            continue
        key, kind, payload = m.group(1), m.group(2).lower(), m.group(3)
        if key in evidence:
            continue
        if kind == "lookup_metric":
            evidence[key] = lookup_metric.invoke({"name": payload})
        elif kind == "calc":
            evidence[key] = calc.invoke({"expression": payload})
        else:
            evidence[key] = search_policy.invoke({"query": payload})
        print(f"  [worker] {key} = {evidence[key]}")
    return {"evidence": evidence}

def solve_node(state: RewooState) -> dict:
    prompt = (
        "Answer using ONLY evidence. Show the math reasoning.\n"
        f"Q: {state['question']}\nPlan:\n{state['plan']}\nEvidence: {state['evidence']}"
    )
    answer = llm.invoke([HumanMessage(content=prompt)]).content
    print("  [solve]", answer[:150], "...")
    return {"answer": answer}

rewoo_builder = StateGraph(RewooState)
rewoo_builder.add_node("plan", plan_node)
rewoo_builder.add_node("worker", worker_node)
rewoo_builder.add_node("solve", solve_node)
rewoo_builder.add_edge(START, "plan")
rewoo_builder.add_edge("plan", "worker")
rewoo_builder.add_edge("worker", "solve")
rewoo_builder.add_edge("solve", END)
rewoo_graph = rewoo_builder.compile()
show_graph(rewoo_graph, "REWOO")

BIZ_Q = "Last month we had 847 signups and conversion rate is 18%. How many paid customers?"
rewoo_out = rewoo_graph.invoke({"question": BIZ_Q, "plan": "", "evidence": {}, "answer": ""})
print("\n=== ANSWER ===\n", rewoo_out["answer"])
```

**What to look at:** Count LLM calls — plan (1) + solve (1) vs ReAct which would need 3+.

# Section 8 — Plan-and-Execute

**Runnable below ↓**

## What it is (2 sentences)

A **planner** LLM writes a multi-step plan; **executor** node(s) run each step with tools; a **replanner** fires if the world changed or a step failed. Production teams use this for SOPs, incident runbooks, and onboarding workflows where step order matters.

## Paper & links

- **Planning inspiration: Wang et al., Plan-and-Solve (ACL 2023):** [Wang, Lei et al.](https://arxiv.org/abs/2305.04091)
- **arXiv (Plan-and-Solve inspiration):** https://arxiv.org/abs/2305.04091
- **BabyAGI:** [https://github.com/yoheinakajima/babyagi](https://github.com/yoheinakajima/babyagi)
- **LangGraph blog:** [https://blog.langchain.com/planning-agents/](https://blog.langchain.com/planning-agents/)
- **LangGraph tutorial:** [https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/plan-and-execute/plan-and-execute.ipynb](https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/plan-and-execute/plan-and-execute.ipynb)

## Paper architecture diagram

![Plan-and-Execute architecture](./assets/patterns/plan_execute_architecture.png)


**LangGraph node graph (matches runnable code below):**

![Plan-and-Execute LangGraph nodes](./assets/patterns/plan_execute_langgraph.png)

## Example use case in this notebook

**SRE incident — 503 errors → 3-step runbook → Slack summary**

## Our LangGraph implementation (you will run this next)


```python
class PlanState(TypedDict):
    incident: str
    steps: list
    current: int
    log: list
    report: str

def planner_node(state: PlanState) -> dict:
    prompt = (
        f"Production incident: {state['incident']}\n"
        "Return exactly 3 numbered action steps for an on-call engineer. "
        "One short sentence per step. Format: 1. ...\n2. ...\n3. ..."
    )
    text = llm.invoke([HumanMessage(content=prompt)]).content
    steps = [ln.strip() for ln in text.splitlines() if ln.strip() and ln.strip()[0].isdigit()]
    steps = steps[:3]
    print("  [planner]", steps)
    return {"steps": steps, "current": 0, "log": []}

def executor_node(state: PlanState) -> dict:
    i = state["current"]
    if i >= len(state["steps"]):
        return {}
    step = state["steps"][i]
    # 👇 Simulate executing the step (in production: call PagerDuty, run kubectl, etc.)
    result = f"Executed step {i+1}: {step} → status OK (simulated)"
    log = list(state["log"]) + [result]
    print(" ", result)
    return {"log": log, "current": i + 1}

def summarize_node(state: PlanState) -> dict:
    prompt = (
        f"Incident: {state['incident']}\n"
        f"Actions taken:\n" + "\n".join(state["log"]) + "\n"
        "Write a 2-sentence incident summary for Slack."
    )
    report = llm.invoke([HumanMessage(content=prompt)]).content
    return {"report": report}

def route_plan(state: PlanState) -> str:
    if state["current"] < len(state.get("steps") or []):
        return "executor"
    return "summarize"

pae_builder = StateGraph(PlanState)
pae_builder.add_node("planner", planner_node)
pae_builder.add_node("executor", executor_node)
pae_builder.add_node("summarize", summarize_node)
pae_builder.add_edge(START, "planner")
pae_builder.add_edge("planner", "executor")
pae_builder.add_conditional_edges("executor", route_plan, {"executor": "executor", "summarize": "summarize"})
pae_builder.add_edge("summarize", END)
pae_graph = pae_builder.compile()
show_graph(pae_graph, "Plan-and-Execute")

INCIDENT = "Checkout API returning HTTP 503 for 15% of requests in us-east-1"
pae_out = pae_graph.invoke({"incident": INCIDENT, "steps": [], "current": 0, "log": [], "report": ""})
print("\n=== SLACK SUMMARY ===\n", pae_out["report"])
```

**What to look at:** Planner runs once; executor loops — matches incident runbook SOPs.

# Section 9 — Tree of Thoughts (ToT)

**Runnable below ↓**

## What it is (2 sentences)

ToT treats reasoning as **tree search**: expand multiple candidate 'thoughts', score each with an LLM judge, prune to top-K, and repeat until a solution is found. Best for problems with a **search space** (puzzles, architecture decisions) — expensive but explores alternatives ReAct misses.

## Paper & links

- **Paper:** [Yao et al., NeurIPS 2023](https://arxiv.org/abs/2305.10601)
- **arXiv:** https://arxiv.org/abs/2305.10601
- **LangGraph tutorial:** [https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/tot/tot.ipynb](https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/tot/tot.ipynb)

## Paper architecture diagram

![Tree of Thoughts (ToT) architecture](./assets/patterns/tot_architecture.png)


**Official LangGraph diagram:**

![Tree of Thoughts (ToT) LangGraph](./assets/patterns/tot_langgraph.png)


**LangGraph node graph (matches runnable code below):**

![Tree of Thoughts (ToT) LangGraph nodes](./assets/patterns/tot_langgraph.png)

## Example use case in this notebook

**Architect — pick best of 3 cloud migration strategies for 50TB PostgreSQL**

## Our LangGraph implementation (you will run this next)


```python
class TotState(TypedDict):
    problem: str
    candidates: list
    scores: list
    best: str

def expand_node(state: TotState) -> dict:
    prompt = (
        f"Problem: {state['problem']}\n"
        "Propose exactly 3 different migration strategies. "
        "One strategy per line, numbered 1-3. Be specific (tool names, downtime, cost tradeoff)."
    )
    text = llm.invoke([HumanMessage(content=prompt)]).content
    candidates = [ln.strip() for ln in text.splitlines() if ln.strip()][:3]
    print("  [expand]")
    for c in candidates:
        print("   -", c[:80])
    return {"candidates": candidates}

def score_node(state: TotState) -> dict:
    scores = []
    for cand in state["candidates"]:
        prompt = (
            f"Score this migration strategy 1-10 for: cost, downtime, team skill fit.\n"
            f"Strategy: {cand}\nReply ONLY with one integer 1-10."
        )
        raw = llm.invoke([HumanMessage(content=prompt)]).content.strip()
        scores.append(int("".join(ch for ch in raw if ch.isdigit())[:2] or "0"))
    print("  [score]", scores)
    return {"scores": scores}

def pick_node(state: TotState) -> dict:
    if not state["candidates"]:
        return {"best": "none"}
    idx = max(range(len(state["scores"])), key=lambda i: state["scores"][i])
    best = state["candidates"][idx]
    print("  [pick]", best[:100])
    return {"best": best}

tot_builder = StateGraph(TotState)
tot_builder.add_node("expand", expand_node)
tot_builder.add_node("score", score_node)
tot_builder.add_node("pick", pick_node)
tot_builder.add_edge(START, "expand")
tot_builder.add_edge("expand", "score")
tot_builder.add_edge("score", "pick")
tot_builder.add_edge("pick", END)
tot_graph = tot_builder.compile()
show_graph(tot_graph, "Tree of Thoughts")

MIGRATION = (
    "We have 50TB PostgreSQL on-prem, 99.9% uptime SLA, team knows AWS not GCP, "
    "budget-conscious, need minimal downtime."
)
tot_out = tot_graph.invoke({"problem": MIGRATION, "candidates": [], "scores": [], "best": ""})
print("\n=== RECOMMENDED STRATEGY ===\n", tot_out["best"])
```

**What to look at:** Three candidates scored independently — explore before committing (vs ReAct's single path).

# Section 10 — Self-Discover

**Runnable below ↓**

## What it is (2 sentences)

Self-Discover **selects reasoning modules** from a library (critical thinking, step-by-step planning, etc.), adapts them to the task, builds a structured JSON plan, then executes it. It is **meta-reasoning** — the model composes its own thinking strategy before solving, ideal for diverse or unfamiliar task types.

## Paper & links

- **Paper:** [Zhou et al., 2024](https://arxiv.org/abs/2402.03620)
- **arXiv:** https://arxiv.org/abs/2402.03620
- **LangGraph tutorial:** [https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/self-discover/self-discover.ipynb](https://github.com/langchain-ai/langgraph/blob/main/docs/docs/tutorials/self-discover/self-discover.ipynb)

## Paper architecture diagram

![Self-Discover architecture](./assets/patterns/self_discover_architecture.png)


**LangGraph node graph (matches runnable code below):**

![Self-Discover LangGraph nodes](./assets/patterns/self_discover_langgraph.png)

## Example use case in this notebook

**Debug RAG — wrong refund answers despite good retrieval scores**

## Our LangGraph implementation (you will run this next)


```python
REASONING_MODULES = [
    "1. Identify failure mode category (retrieval vs generation vs chunking)",
    "2. Break into sub-problems and test each hypothesis",
    "3. Compare against a known-good baseline",
    "4. Check edge cases in policy text",
    "5. Step-by-step debugging plan with metrics",
]

class DiscoverState(TypedDict):
    question: str
    selected: str
    adapted: str
    structure: str
    answer: str

def select_node(state: DiscoverState) -> dict:
    prompt = (
        f"Pick 2-3 reasoning modules for this debugging task:\n{REASONING_MODULES}\n"
        f"Task: {state['question']}\nList chosen modules and why."
    )
    selected = llm.invoke([HumanMessage(content=prompt)]).content
    print("  [select]", selected[:120], "...")
    return {"selected": selected}

def adapt_node(state: DiscoverState) -> dict:
    prompt = f"Adapt modules to RAG debugging:\n{state['selected']}\nTask: {state['question']}"
    adapted = llm.invoke([HumanMessage(content=prompt)]).content
    print("  [adapt]", adapted[:100], "...")
    return {"adapted": adapted}

def structure_node(state: DiscoverState) -> dict:
    prompt = f"Turn into numbered checklist a team can follow:\n{state['adapted']}"
    structure = llm.invoke([HumanMessage(content=prompt)]).content
    print("  [structure]", structure[:100], "...")
    return {"structure": structure}

def reason_node(state: DiscoverState) -> dict:
    prompt = (
        f"Problem: {state['question']}\n"
        f"Follow this plan:\n{state['structure']}\n"
        "Give root-cause hypotheses + top 3 fixes ranked by impact."
    )
    answer = llm.invoke([HumanMessage(content=prompt)]).content
    print("  [reason]", answer[:150], "...")
    return {"answer": answer}

discover_builder = StateGraph(DiscoverState)
discover_builder.add_node("select", select_node)
discover_builder.add_node("adapt", adapt_node)
discover_builder.add_node("structure", structure_node)
discover_builder.add_node("reason", reason_node)
discover_builder.add_edge(START, "select")
discover_builder.add_edge("select", "adapt")
discover_builder.add_edge("adapt", "structure")
discover_builder.add_edge("structure", "reason")
discover_builder.add_edge("reason", END)
discover_graph = discover_builder.compile()
show_graph(discover_graph, "Self-Discover")

RAG_Q = (
    "Our RAG chatbot confidently answers refund policy questions incorrectly. "
    "Retrieval scores look fine. What should we investigate?"
)
disc_out = discover_graph.invoke({
    "question": RAG_Q, "selected": "", "adapted": "", "structure": "", "answer": "",
})
print("\n=== DIAGNOSIS + FIXES ===\n", disc_out["answer"])
```

**What to look at:** Linear graph but **meta-reasoning** happens before the final answer.

# Section 11 — Bonus patterns

These patterns are reference-only — diagrams + paper links below.

### Graph of Thoughts (GoT)

GoT generalizes ToT from a tree to an arbitrary **DAG** — thoughts can merge, backtrack, and combine from different branches. Use when sub-problems solved independently need to be **fused** into one answer (e.g., multi-document synthesis).

- **Paper:** [Besta et al., 2023](https://arxiv.org/abs/2308.09687)

![Graph of Thoughts (GoT)](./assets/patterns/got_architecture.png)

### LATS (Language Agent Tree Search)

LATS combines **Monte Carlo Tree Search** with LLM reflection — select a node, expand parallel actions, score with reflection, backpropagate rewards up the tree. It unifies ReAct, Reflexion, ToT, and plan-and-execute into one search algorithm — powerful but the highest cost pattern.

- **Paper:** [Zhou et al., 2023](https://arxiv.org/abs/2310.04406)

![LATS (Language Agent Tree Search)](./assets/patterns/lats_architecture.png)

### Self-RAG

Self-RAG trains/reflection-tokens decide **when to retrieve**, whether the generation is **grounded**, and whether the answer is **useful** — retrieve again or ship. Your Module 10 capstone (`grade_documents` → `rewrite_query` → `check_groundedness`) is a domain-specific Self-RAG graph.

- **Paper:** [Asai et al., 2023](https://arxiv.org/abs/2310.11511)

![Self-RAG](./assets/patterns/self_rag_architecture.png)

### LLM Compiler

LLM Compiler plans a **DAG of tool calls** and executes independent nodes **in parallel**, joining results as dependencies resolve — like a task scheduler for LLM tools. Fastest pattern when tools don't depend on each other's output sequentially (e.g., fetch weather + stock price + news simultaneously).

- **Paper:** [Kim et al., 2023](https://arxiv.org/abs/2312.04511)

![LLM Compiler](./assets/patterns/llm_compiler_architecture.png)

## 15 interview questions (with crisp answers)

1. **ReAct in one sentence?** Interleave reasoning traces with tool actions so the model can gather external facts mid-task. [2210.03629](https://arxiv.org/abs/2210.03629)

2. **Reflection vs Reflexion?** Reflection = inner critic, no tools ([Self-Refine, 2303.17651](https://arxiv.org/abs/2303.17651)). Reflexion = critic **after** environment/tool feedback + episodic verbal memory ([2303.11366](https://arxiv.org/abs/2303.11366)).

3. **CRITIC vs Reflexion?** Both use tools. CRITIC focuses on **verify-then-correct** with explicit tool grounding; ablation shows self-critique alone fails. [2305.11738](https://arxiv.org/abs/2305.11738)

4. **When REWOO over ReAct?** 5+ tool steps where observations need not be re-read by LLM each time — plan once, execute, solve once. [2305.18323](https://arxiv.org/abs/2305.18323)

5. **Plan-and-Execute vs REWOO?** PAE replans when world changes ([Plan-and-Solve inspiration, 2305.04091](https://arxiv.org/abs/2305.04091)); REWOO assumes plan stays valid after tool runs.

6. **ToT vs CoT?** CoT = one chain. ToT = **multiple branches**, scored and pruned. [2305.10601](https://arxiv.org/abs/2305.10601)

7. **LATS vs ToT?** LATS adds MCTS backpropagation + reflection — stronger, far more expensive. [2310.04406](https://arxiv.org/abs/2310.04406)

8. **Self-Discover vs Self-RAG?** Self-Discover composes reasoning **modules** for any task. Self-RAG reflects on **retrieval** need/quality for RAG. [2402.03620](https://arxiv.org/abs/2402.03620) / [2310.11511](https://arxiv.org/abs/2310.11511)

9. **GoT vs ToT?** ToT is a tree. GoT allows **merging** thoughts from different branches (DAG). [2308.09687](https://arxiv.org/abs/2308.09687)

10. **LLM Compiler use case?** Independent tools as a DAG — parallel execution, minimum latency. [2312.04511](https://arxiv.org/abs/2312.04511)

11. **Can LLMs self-correct without tools?** Often no — [Huang et al.](https://arxiv.org/abs/2310.01798) shows intrinsic self-correction is unreliable; external feedback (CRITIC) is key.

12. **Safety cap for loops?** Always `MAX_ITERATIONS` / max attempts — infinite loops burn budget.

13. **LangGraph advantage?** Explicit graph = visible loops, conditional edges, HITL — not a hidden while-loop.

14. **What pattern is Module 10 capstone?** Domain-specific **Reflexion/Self-RAG**: grade retrieval → rewrite query → regenerate → escalate.

15. **Production default?** Start `create_react_agent`; add Reflection/Reflexion/REWOO only when evals prove the baseline fails.
